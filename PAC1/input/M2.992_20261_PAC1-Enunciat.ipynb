### 1.2. Construcció i comparació de dos models nuls (1,5 punts)

Volem determinar quines propietats de la xarxa real es poden considerar veritablement característiques de la seva estructura. Per fer-ho, construeix dos models nuls diferents:

```python
# REFERÈNCIES BIBLIOGRÀFIQUES UTILITZADES
# Casas Roma, J. (2023). Mètriques i propietats dels grafs. 1a ed.
# Barcelona: FUOC. PID_00286317, §3.2 (distàncies), §3.3 (centralització
# de grau, eq. 13) i §3.4 (clustering mitjà, eq. 16).
# Pérez Solà, C. (2023). Generació de grafs sintètics. 1a ed.
# Barcelona: FUOC. PID_00286323, §1.1 (G(n,m)) i §2.1 (graus fixats).
# NetworkX Developers (2026). Documentació de NetworkX 3.7.
# https://networkx.org/documentation/stable/reference/generated/networkx.generators.random_graphs.gnm_random_graph.html
# https://networkx.org/documentation/stable/reference/algorithms/generated/networkx.algorithms.swap.double_edge_swap.html
# Consulta: 04/10/2026. Els intercanvis d'arestes són una alternativa
# per a grafs simples amb graus fixats, no el model de caps dels apunts.

from pathlib import Path
from IPython.display import display, Markdown
import networkx as nx
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

candidats = [Path("network_real.edgelist"), Path("input/network_real.edgelist"),
             Path("PAC1/input/network_real.edgelist")]
fitxer = next((p for p in candidats if p.is_file()), None)
if fitxer is None:
    raise FileNotFoundError("No s'ha trobat network_real.edgelist")
G = nx.read_edgelist(fitxer, nodetype=int, data=False, create_using=nx.Graph())
assert not G.is_directed() and nx.number_of_selfloops(G) == 0
n, m = G.number_of_nodes(), G.number_of_edges()
graus_reals = dict(G.degree())
etiquetes = list(G)
metriques_12 = ["Clustering mitjà", "Camí més curt mitjà",
                "Diàmetre", "Centralització de grau"]
models = ["Nul 1: G(n,m)", "Nul 2: graus fixats"]
colors = ["tab:blue", "tab:orange"]
rng = np.random.default_rng(20261)

def llavor():
    return int(rng.integers(0, 2**32 - 1))

def mesures_12(H):
    # Mateix criteri per a totes les xarxes: distàncies a la component major.
    # Es registra la mida per evitar ocultar una possible desconnexió.
    component = max(nx.connected_components(H), key=len)
    Hc = H.subgraph(component)
    nc = len(Hc)
    suma, diametre = 0, 0
    for _, distancies in nx.all_pairs_shortest_path_length(Hc):
        suma += sum(distancies.values())
        diametre = max(diametre, max(distancies.values()))
    graus = np.array([d for _, d in H.degree()])
    centralitzacio = np.sum(graus.max() - graus) / ((len(H) - 1) * (len(H) - 2))
    return {
        "Clustering mitjà": nx.average_clustering(H),
        "Camí més curt mitjà": suma / (nc * (nc - 1)) if nc > 1 else 0.0,
        "Diàmetre": diametre,
        "Centralització de grau": centralitzacio,
        "Nodes component major": nc,
        "Connexa": nx.is_connected(H),
    }

def comprova_12(H, conserva_graus=False):
    assert set(H) == set(G), "Han canviat els nodes"
    assert H.number_of_edges() == m, "Han canviat les arestes"
    assert not H.is_multigraph() and nx.number_of_selfloops(H) == 0
    if conserva_graus:
        assert dict(H.degree()) == graus_reals, "Han canviat els graus"

def recableja_12(H, nombre):
    # Cada intercanvi acceptat conserva el grau dels quatre extrems.
    # No s'imposa connexitat: seria una restricció addicional del model nul.
    nx.double_edge_swap(H, nswap=nombre, max_tries=100 * nombre, seed=llavor())
    comprova_12(H, conserva_graus=True)

real_12 = mesures_12(G)

# Diagnòstic de recablejat: cinc cadenes amb llavors diferents.
# 20*m intercanvis és una elecció empírica; no prova mostreig uniforme exacte.
pressupostos = [0, 1, 5, 10, 20]
pilot = []
for repeticio in range(5):
    H = G.copy()
    anterior = 0
    for factor in pressupostos:
        if factor > anterior:
            recableja_12(H, (factor - anterior) * m)
        pilot.append({"Repetició": repeticio, "Intercanvis / m": factor,
                      **mesures_12(H)})
        anterior = factor
pilot_12 = pd.DataFrame(pilot)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
for ax, metrica in zip(axes, metriques_12[:2]):
    resum = pilot_12.groupby("Intercanvis / m")[metrica].agg(["mean", "std"])
    ax.errorbar(resum.index, resum["mean"], yerr=resum["std"], marker="o", capsize=4)
    ax.set(xlabel="Intercanvis acceptats / nombre d'arestes", ylabel=metrica,
           title="Diagnòstic del recablejat: mitjana ± DE (5 cadenes)")
    ax.grid(alpha=0.2)
plt.show()
print("Pilot: comprova visualment que les mètriques s'estabilitzen entre 10m i 20m.")

# Realitzacions independents (llavors diferents; cada recablejat comença a G).
# Mínim 100; afegim lots de 50 fins a assolir la precisió o arribar a 300.
# Criteri: semiamplada IC aproximat 95% de la MITJANA <= 2% de la mitjana,
# i diferència relativa entre les mitjanes de les dues meitats <= 2%.
# Aquests IC mesuren l'error Monte Carlo, no la dispersió entre grafs.
MIN_REALITZACIONS, MAX_REALITZACIONS, LOT = 100, 300, 50
TOLERANCIA = 0.02
INTERCANVIS_PER_ARESTA = 20
registres = []

def diagnostics_12(dades):
    files = []
    for model in models:
        sub = dades.loc[dades["Model"] == model]
        r = len(sub)
        for metrica in metriques_12:
            valors = sub[metrica].to_numpy(dtype=float)
            mitjana = valors.mean()
            escala = max(abs(mitjana), 1e-12)
            semi_ic = 1.96 * valors.std(ddof=1) / np.sqrt(r)
            canvi = abs(valors[:r // 2].mean() - valors[r // 2:].mean()) / escala
            files.append({"Model": model, "Mètrica": metrica, "R": r,
                          "Semi-IC / mitjana": semi_ic / escala,
                          "Diferència meitats / mitjana": canvi,
                          "Estable": semi_ic / escala <= TOLERANCIA and canvi <= TOLERANCIA})
    return pd.DataFrame(files)

for r in range(1, MAX_REALITZACIONS + 1):
    H1 = nx.gnm_random_graph(n, m, seed=llavor())
    H1 = nx.relabel_nodes(H1, dict(enumerate(etiquetes)))
    comprova_12(H1)
    H2 = G.copy()
    recableja_12(H2, INTERCANVIS_PER_ARESTA * m)
    for model, H in zip(models, [H1, H2]):
        registres.append({"Model": model, "Realització": r, **mesures_12(H)})
    if r % LOT == 0:
        print(f"Completades {r} realitzacions per model.", flush=True)
    if r >= MIN_REALITZACIONS and r % LOT == 0:
        resultats_12 = pd.DataFrame(registres)
        diagnostic_12 = diagnostics_12(resultats_12)
        if diagnostic_12["Estable"].all():
            break

display(Markdown("### Verificació i precisió Monte Carlo"))
print(f"{r} realitzacions per model; nodes, arestes i simplicitat verificats en totes.")
print("Model nul 2: grau de cada node verificat en totes les realitzacions.")
print("Distàncies: component connexa major en tots els casos; no es descarten grafs.")
display(resultats_12.groupby("Model").agg(
    Realitzacions=("Realització", "count"),
    Connexes=("Connexa", "sum"),
    Mínim_nodes_component_major=("Nodes component major", "min")))
display(diagnostic_12.style.format({"Semi-IC / mitjana": "{:.2%}",
                                   "Diferència meitats / mitjana": "{:.2%}"}))
if not diagnostic_12["Estable"].all():
    print("El límit de simulacions s'ha assolit sense satisfer tots els criteris; cal augmentar-lo.")

# Taula de comparació: DE entre realitzacions i percentils empírics del model.
comparacio = []
for model in models:
    for metrica in metriques_12:
        valors = resultats_12.loc[resultats_12["Model"] == model, metrica].to_numpy()
        q025, q975 = np.quantile(valors, [0.025, 0.975])
        comparacio.append({"Model": model, "Mètrica": metrica,
                           "Xarxa real": real_12[metrica], "Mitjana nul": valors.mean(),
                           "DE nul": valors.std(ddof=1), "Percentil 2,5": q025,
                           "Percentil 97,5": q975})
comparacio_12 = pd.DataFrame(comparacio)
display(comparacio_12.style.format({c: "{:.4f}" for c in comparacio_12.columns[2:]}))

fig, axes = plt.subplots(2, 2, figsize=(11, 8), constrained_layout=True)
for ax, metrica in zip(axes.flat, metriques_12):
    mostres = [resultats_12.loc[resultats_12["Model"] == model, metrica] for model in models]
    caixes = ax.boxplot(mostres, tick_labels=["Nul 1", "Nul 2"], patch_artist=True)
    for caixa, color in zip(caixes["boxes"], colors):
        caixa.set(facecolor=color, alpha=0.45)
    ax.axhline(real_12[metrica], color="crimson", linestyle="--", label="Xarxa real")
    ax.set(title=metrica, ylabel="Valor")
    ax.grid(axis="y", alpha=0.2)
    ax.legend()
fig.suptitle(f"Xarxa real i models nuls ({r} realitzacions per model)")
plt.show()

# Convergència: mitjanes acumulades i IC aproximats 95% de cada mitjana.
fig, axes = plt.subplots(2, 2, figsize=(11, 8), constrained_layout=True)
for ax, metrica in zip(axes.flat, metriques_12):
    for model, color in zip(models, colors):
        valors = resultats_12.loc[resultats_12["Model"] == model, metrica].reset_index(drop=True)
        comptador = np.arange(1, len(valors) + 1)
        mitjanes = valors.expanding().mean()
        semi_ic = 1.96 * valors.expanding().std(ddof=1) / np.sqrt(comptador)
        ax.plot(comptador, mitjanes, color=color, label=model)
        ax.fill_between(comptador, mitjanes - semi_ic, mitjanes + semi_ic, color=color, alpha=0.2)
    ax.set(title=metrica, xlabel="Nombre de realitzacions", ylabel="Mitjana acumulada")
    ax.grid(alpha=0.2)
    ax.legend(fontsize=8)
fig.suptitle("Estabilitat de les estimacions: mitjana acumulada i IC aproximat 95%")
plt.show()

display(Markdown("""### Justificació dels models i de les mètriques
* **Nul 1, G(n,m):** conserva exactament els nodes i les arestes i, per tant,
  la densitat i el grau mitjà. Aleatoritza qui interactua amb qui i no conserva
  la seqüència de graus, els triangles ni les agrupacions de la xarxa original.
* **Nul 2, intercanvis d'arestes:** conserva el grau de cada individu i,
  per tant, la distribució de graus, els nodes i les arestes. Aleatoritza les
  parelles d'interacció; no imposa triangles, comunitats ni correlacions entre
  graus. Aquestes estructures poden aparèixer per atzar, però no es fixen.
  El model de configuració dels apunts pot produir multiarestes i bucles;
  simplificar-lo canviaria els graus. Per això aquí s'utilitza recablejat
  sobre grafs simples. El pilot comprova empíricament l'estabilització,
  però no demostra una barreja perfecta ni uniformitat exacta del mostreig.
* **Clustering mitjà:** mesura el tancament local de triangles.
  **Camí mitjà:** mesura la separació típica entre individus.
  **Diàmetre:** mesura la major separació mínima, complementant la mitjana.
  **Centralització de grau:** quantifica la concentració de connexions
  respecte del node de grau màxim (equació 13 dels apunts).
  Ha de quedar exactament fixada al nul 2; al nul 1 pot variar.
* **Nombre de realitzacions:** es determina amb els criteris de precisió
  mostrats a la taula. Les corbes acumulades permeten valorar l'estabilitat
  i la reducció de l'error Monte Carlo. Les bandes de convergència són IC
  de la mitjana; els boxplots mostren la variabilitat entre xarxes.
"""))
display(Markdown("### Interpretació de la comparació"))
for fila in comparacio:
    valor = fila["Xarxa real"]
    baix, alt = fila["Percentil 2,5"], fila["Percentil 97,5"]
    posicio = "per sota" if valor < baix else "per sobre" if valor > alt else "dins"
    print(f"{fila['Model']} · {fila['Mètrica']}: real={valor:.4f}; "
          f"rang central empíric 95%=[{baix:.4f}, {alt:.4f}]; {posicio} del rang.")
print("Una diferència que persisteix al nul 2 no s'explica només pels graus. "
      "Els percentils són una comparació descriptiva, no una prova formal amb correcció múltiple.")
```
